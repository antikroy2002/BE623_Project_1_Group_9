```python
seq = ""

for line in open("sequence.fasta"):
    if not line.startswith(">"):
        seq += line.strip()

cds = seq[363:1728]
print(cds)
print("total nucleotide :" , len(cds))
print("number of codons:", len(cds)/3)
```

## For the Myc gene, CTG codes for leucine, for this gene, CTG acts as a noncanonical start codon instead of the usual ATG, which codes for methionine. GenBank shows M as the first amino acid because cellular translation always uses methionine-charged tRNA therefore the original sequence always contain methionine as first amino acid.
## Therefore, the Python-translated script will show Leucine regardless of what the first amino acid is, as it is actually added after translation.

```python
gcode = {"TTT":"F", "TTC":"F", "TTA":"L", "TTG":"L",
    "TCT":"S", "TCC":"S", "TCA":"S", "TCG":"S",
    "TAT":"Y", "TAC":"Y", "TAA":"*", "TAG":"*",
    "TGT":"C", "TGC":"C", "TGA":"*", "TGG":"W",

    "CTT":"L", "CTC":"L", "CTA":"L", "CTG":"L",
    "CCT":"P", "CCC":"P", "CCA":"P", "CCG":"P",
    "CAT":"H", "CAC":"H", "CAA":"Q", "CAG":"Q",
    "CGT":"R", "CGC":"R", "CGA":"R", "CGG":"R",

    "ATT":"I", "ATC":"I", "ATA":"I", "ATG":"M",
    "ACT":"T", "ACC":"T", "ACA":"T", "ACG":"T",
    "AAT":"N", "AAC":"N", "AAA":"K", "AAG":"K",
    "AGT":"S", "AGC":"S", "AGA":"R", "AGG":"R",

    "GTT":"V", "GTC":"V", "GTA":"V", "GTG":"V",
    "GCT":"A", "GCC":"A", "GCA":"A", "GCG":"A",
    "GAT":"D", "GAC":"D", "GAA":"E", "GAG":"E",
    "GGT":"G", "GGC":"G", "GGA":"G", "GGG":"G"}

print("Number of codons:", len(gcode))
```

```python
protein = ""
for i in range(0, len(cds), 3):
    codon = cds[i:i+3]
    protein= protein+ gcode[codon]

print("Protein length:", len(protein))
print("First 30 residues:", protein[:30])
```
